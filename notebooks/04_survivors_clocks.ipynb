# Script 04 – Epigenetic clocks in the 281 matched survivors

Downloads the SJLIFE1 file (about 7 GB) in Google's cloud, copies out only the 281 survivors chosen in script 02, and computes their clocks. Takes roughly 45–90 minutes in total, mostly download and unpacking.

**Before you start:** Runtime → Change runtime type → **R** → Save. Run the cells one at a time, top to bottom, and keep this tab open.

## 1. Get a fresh copy of the project code

```r
setwd("/content")
unlink("/content/sjlife-epigenetic-age", recursive = TRUE)
system("git clone https://github.com/Mrn-selyn-commits/sjlife-epigenetic-age.git")
setwd("/content/sjlife-epigenetic-age")
list.files("scripts")
```

## 2. Install the packages (15–30 minutes; ends with a row of TRUE)

```r
source("scripts/00_setup.R")
```

## 3. Download, extract the 281 survivors and compute their clocks
Progress messages appear as it goes. If the session disconnects, check with `list.files("data/raw")` before starting over: a finished download or extract is reused.

```r
source("scripts/04_survivors_clocks.R")
```

## 4. Look at the result

```r
cat(readLines("results/04_survivors_clock_summary.txt"), sep = "\n")
```
