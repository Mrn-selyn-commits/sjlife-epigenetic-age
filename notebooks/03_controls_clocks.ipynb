# Script 03 – Epigenetic clocks in the 282 controls

Runs `scripts/03_controls_clocks.R` in Google Colab, so the 1 GB GEO download happens in the cloud, not on your data bundle.

**Before you start:** Runtime → Change runtime type → **R** → Save.

Run the cells one at a time, top to bottom (click ▶ on each). Wait for one to finish before starting the next.

## 1. Get the project code from GitHub

```r
system("git clone https://github.com/Mrn-selyn-commits/sjlife-epigenetic-age.git")
setwd("sjlife-epigenetic-age")
list.files("scripts")
```

## 2. Install the packages
The first time this takes about 15–30 minutes. Warnings in red are normal; only stop if it ends with *Error*.

```r
source("scripts/00_setup.R")
```

## 3. Download the controls' methylation data and compute the clocks
About 10–20 minutes.

```r
source("scripts/03_controls_clocks.R")
```

## 4. Look at the result

```r
cat(readLines("results/03_controls_clock_summary.txt"), sep = "\n")
IRdisplay::display_png(file = "results/03_controls_clock_vs_age.png")
```

## 5. Save the results before closing
Colab deletes everything when the session ends.

1. Click the **folder icon** on the left.
2. Open `sjlife-epigenetic-age/results`.
3. For each of the three `03_` files: click ⋮ → **Download**.
4. Also download `sjlife-epigenetic-age/data/processed/03_controls_clock_ages.csv` and keep it on your computer (not on GitHub).
5. Upload the three `results/03_` files to the `results` folder on GitHub.
